## Setup

```python
from pathlib import Path
import json, csv, re, collections, itertools, heapq, statistics, datetime, zoneinfo, math

DATA_DIR = Path('data')
if not DATA_DIR.exists():
    DATA_DIR = Path('python_de_practice/data')
assert DATA_DIR.exists(), 'Run the dataset setup notebook first, or put the data/ folder next to this notebook.'

def load(name):
    with (DATA_DIR / f'{name}.json').open(encoding='utf-8') as f:
        return json.load(f)

def read_jsonl(name):
    with (DATA_DIR / f'{name}.jsonl').open(encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
print('Data folder:', DATA_DIR.resolve())
```

### Q21 — Top 5 spending customers

**Interview scenario:** You are building an online retail order-processing platform. A teammate asks you to implement the following production-oriented transformation using plain Python (standard-library collections and file processing).

**Requirements:**
- Define a deterministic ordering or tie-breaker where records can have equal values; preserve the required row granularity.
- Use numeric calculations and keep monetary comparisons precise enough for cents.
- Implement using Python and the provided file-loading helpers; do not assume the dataset is already cleaned.

**Available datasets and columns:**
- **`orders`**: order_id, customer_id, order_date, status, amount

**Expected output:** Return or display the records, summary table, or derived values described in the task. Include the requested fields; when the task is a count, return the numeric count (zero is valid).

**Solution:**


```python
totals = collections.defaultdict(float)
for o in load('orders'):
    if isinstance(o.get('amount'), (int,float)) and o.get('customer_id') is not None:
        totals[o['customer_id']] += o['amount']
result = sorted(totals.items(), key=lambda x: (-x[1], x[0]))[:5]
print(result)
```

### Q74 — Retry transient failures

**Requirements:**
- Use the supplied columns as the source of truth, preserve the requested output granularity, and make the result reproducible.
- Implement using Python and the provided file-loading helpers; do not assume the dataset is already cleaned.

**Available datasets and columns:**
- **`pipeline_logs`**: log_id, pipeline, level, duration_ms, timestamp, message

**Expected output:** Return or display the records, summary table, or derived values described in the task. Include the requested fields; when the task is a count, return the numeric count (zero is valid).

**Solution:**


```python
attempts=0
def unreliable():
    global attempts
    attempts+=1
    if attempts<3:raise TimeoutError('transient timeout')
    return 'ok'
def retry(fn,max_attempts=3):
    for attempt in range(max_attempts):
        try:return fn()
        except TimeoutError:
            if attempt==max_attempts-1:raise
print(retry(unreliable),attempts)
```
