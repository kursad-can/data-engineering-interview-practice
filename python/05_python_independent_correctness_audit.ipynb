# Python Data Engineering — 80-question correctness audit

Runs the original solution for each question and a separately authored reference assertion. **PASS** means the defined contract passes on the supplied dataset, not universal correctness. No test is silently marked successful. Outputs `python_correctness_final_report.csv`.

````python
import contextlib, io, ast, traceback
from pathlib import Path
import json, csv, re, collections, itertools, heapq, statistics, datetime, zoneinfo, math

DATA_DIR = Path('data')
if not DATA_DIR.exists():
    DATA_DIR = Path('python_de_practice/data')
assert DATA_DIR.exists(), 'Run the dataset setup notebook first, or put the data/ folder next to this notebook.'

def load(name):
    with (DATA_DIR / f'{name}.json').open(encoding='utf-8') as f:
        return json.load(f)

def read_jsonl(name):
    with (DATA_DIR / f'{name}.jsonl').open(encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)
print('Data folder:', DATA_DIR.resolve())

AUDIT=[]
````
